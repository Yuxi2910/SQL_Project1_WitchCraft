## Q3. The trial
Let's take a look at the information on the trial (table).

```python
%%sql
SELECT *
FROM trial
LIMIT 3;
```

```python
%config SqlMagic.displaylimit = 0
```

```python
%%sql
DESCRIBE trial;
```

```python
#a) What are the average and maximum numbers of male and female accusers?
%%sql
SELECT
    AVG(male_accusers) AS avg_male_accusers,
    MAX(male_accusers) AS max_male_accusers,
    AVG(female_accusers) AS avg_female_accusers,
    MAX(female_accusers) AS max_female_accusers
FROM trial;
```

```python
# b) Count the number of sentences by sentence type. List them in a table (in descending order),
# excluding missing values. Rename the column headings to something sensible.
%%sql
SELECT
  sentence AS Sentence_Type,
  COUNT(*) AS N_Sentences
FROM trial
WHERE sentence IS NOT NULL
GROUP BY Sentence_Type
ORDER BY N_Sentences DESC;
```

```python
# c) Do the number of accusers matter for the verdict? Calculate the total number of accusers (male + female)
# for each trial, then compare the average total number of accusers by the type of verdict.
# Again make sure the table is sorted and the headings make sense.

%%sql
SELECT
  verdict,
  AVG(male_accusers + female_accusers) AS average_total_accusers
FROM trial
GROUP BY verdict
ORDER BY average_total_accusers DESC;
```

The average number of accusers varies significantly depending on the verdict. Trials resulting in a 'Guilty' verdict had the highest average number of total accusers at 5.8.
Trials with a 'Not guilty' verdict showed an average of 4.56 accusers, ranking second.
Trials concluding with a 'None' verdict had the lowest average number of accusers, approximately 0.2, most accusers are being given a clear verdict.

## Q4. Tortured Truth



```python
%%sql
SELECT *
FROM trial
LIMIT 10;
```

```python
%%sql
SELECT *
FROM confession
LIMIT 3;
```

```python
# a) Left join the trial and confession tables. For what share of trials does the database record confessions?
# Create a results table with the number of all trials, the number of confessions, and the share of trials with
# confessions recorded.

%%sql
SELECT
  COUNT(DISTINCT T.trialref) AS N_trials,
  COUNT(C.confessionref) AS N_confessions,
  ROUND((COUNT(C.confessionref) / COUNT(DISTINCT T.trialref)) *100, 2) AS Share_of_Trials_with_Confessions_Percent
FROM trial AS T
LEFT JOIN confession AS C
  ON T.trialref = C.trialref;
```

This suggests that most trials are not matched with a confession, people who were accused did not confess of doing witchcraft.

```python
# b) Only a small number of trials have records of torture. Is there a higher share of confessions among trials
# with records of torture than trials without such record? Hint: You will need to merge on the confession table
# and use a CASE WHEN statement or WHERE conditions with proper parentheses to categorize trials by whether they
# have torture records.

%%sql
SELECT
  CASE
    WHEN T.pretrialnotes LIKE '%torture%' THEN 'With Torture Record'
    ELSE 'Without Torture Record'
  END AS torture_category,
  COUNT(DISTINCT T.trialref) AS total_trials,
  COUNT(C.confessionref) AS num_confessions,
  ROUND((COUNT(C.confessionref) * 100.0) / COUNT(DISTINCT T.trialref), 2) AS percentage_confessions
FROM trial AS T
LEFT JOIN confession AS C
  ON T.trialref = C.trialref
GROUP BY
  torture_category;
```

Thought of process:
- Start from trial again.

- LEFT JOIN to torture (any row ⇒ “has torture record”).

- LEFT JOIN to confession.

- Use CASE WHEN on the joined torture table to split trials into:

    - “Torture record”

    - “No torture record”

- For each group, compute:

    - total number of trials

    - number with confessions

    - share with confessions

*   The analysis categorized trials into two groups: "With Torture Record" and "Without Torture Record," based on whether the `pretrialnotes` column contained the word 'torture'.
*   For each category, the total number of unique trials, the total number of confessions, and the percentage of confessions relative to trials were calculated.
*   The percentage of confessions was determined by dividing the count of confessions by the count of distinct trials for each category, rounded to two decimal places.

Yes, the confession rate is much higher 84.38% with torture, whereas this rate is much lower when torture was not applied (28.19).
